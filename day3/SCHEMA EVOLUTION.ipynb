```python
from pyspark.sql.functions import *
data = [(1, "Virat"), (2, 'Sachin')]
schema = "id int, name string"
df = spark.createDataFrame(data, schema)
display(df)
```

```python
df.write.saveAsTable("dev.srinu.cricket_data")
```

```python
data=([(3,"Rohit","Batsman"),(4,'Rahul',"Bowler")])
schema="emp_id int, emp_name string, role string"
df=spark.createDataFrame(data,schema)
```

```python
df.write.mode("append").option("mergeSchema", "true").saveAsTable("dev.srinu.cricket_data")
```

```python
%sql
select * from dev.srinu.cricket_data
```

```python
%sql
describe history dev.srinu.cricket_data
```
